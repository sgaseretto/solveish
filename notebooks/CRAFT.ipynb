## Project Setup

This CRAFT notebook auto-executes when any notebook in this directory is opened.

It sets up the `my_pkg/` reuse workflow:
- Ensures `my_pkg/` is on `sys.path` so `from my_pkg.module import ...` works
- Shows a status banner while loading

### The my_pkg Reuse Workflow

1. Mark useful cells in any notebook with `#| export`
2. Add `#| default_exp module_name` to set the module name
3. Save the notebook → `my_pkg/module_name.py` is auto-generated
4. Import from `my_pkg.module_name` in any other notebook

````python
# CRAFT auto-setup: ensure my_pkg is importable
import sys
from pathlib import Path

_pkg_path = str(Path.cwd())
if _pkg_path not in sys.path:
    sys.path.insert(0, _pkg_path)
    print(f'[CRAFT] Added {_pkg_path} to sys.path')
else:
    print(f'[CRAFT] sys.path already configured')

# Check if my_pkg exists and list available modules
_pkg_dir = Path.cwd() / 'my_pkg'
if _pkg_dir.exists():
    modules = [f.stem for f in _pkg_dir.glob('*.py') if f.stem != '__init__']
    if modules:
        print(f'[CRAFT] Available my_pkg modules: {', '.join(modules)}')
    else:
        print('[CRAFT] my_pkg/ exists but has no modules yet')
else:
    print('[CRAFT] No my_pkg/ directory yet \u2014 save a notebook with #| default_exp to create one')
````
