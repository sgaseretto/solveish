# pshnb: Persistent Shell for Notebooks

This guide explains how to use shell commands in Dialeng using pshnb (persistent shell for notebooks).

## Overview

pshnb provides a persistent shell session within your notebook, allowing you to:
- Run bash commands directly from code cells using `%bash` magic
- Maintain shell state across multiple cells
- Use Python variable expansion with `@{variable}` syntax

## Installation

pshnb is automatically installed with Dialeng. If you need to install it separately:

```bash
pip install pshnb
```

## Basic Usage

There are three ways to run shell commands in Dialeng:

1. **`!command`** - Simple one-liner (works with flags like `-la`)
2. **`%bash command`** - pshnb line magic (simple commands only)
3. **`%%bash`** - Cell magic for multi-line scripts

### Using `!` prefix (Recommended for one-liners)

````python
# List files with flags - use ! prefix
!ls -la
````

````python
# Check current working directory - simple command, %bash works
%bash pwd
````

````python
# Show system information - has flag, use !
!uname -a
````

## Multi-line Commands with %%bash

For multi-line shell scripts, use `%%bash` cell magic. This is also the best way to run commands with flags.

**Important:** `%%bash` must be on the very first line of the cell (no comments before it).

````python
%%bash
# This comment is inside the bash script, which is fine
echo 'Hello from bash!'
echo "Current date: $(date)"
echo "User: $USER"
````

## Variable Expansion: @{var} Syntax

One of pshnb's powerful features is variable expansion from your Python namespace. Define variables in Python, then reference them in shell commands using `@{variable_name}`:

````python
# Define Python variables
name = "World"
count = 3
files = "*.py"
````

````python
# Use Python variables in shell commands (use single quotes to avoid history expansion)
%bash echo 'Hello, @{name}!'
````

More complex variable usage with `%%bash` (note: `%%bash` must be first line):

````python
%%bash
for i in $(seq 1 @{count}); do
    echo "Iteration $i"
done
````

````python
# Use in file operations
%bash ls @{files} 2>/dev/null || echo 'No Python files found'
````

## Shell Cell Type

If shell cells are enabled in Settings, Dialeng provides a dedicated **Shell** cell type for bash-only execution. Shell cells:
- Have bash syntax highlighting
- Execute directly via pshnb
- Support Safe Mode validation
- Get a fresh shell session per cell (no persistent state)

To create a shell cell, click the **+ Shell** button or change a cell's type to "shell" in the dropdown.

## Timeout Settings

By default, commands timeout after 2 seconds. You can increase the timeout by first configuring pshnb, then running your command:

````python
# Set timeout to 10 seconds (run this first)
%bash -t 10
````

Now run a longer command (uses the 10s timeout from above):

````python
%%bash
sleep 2
echo 'Done sleeping!'
````

## Important Notes on Syntax

### Cell Magic Must Be First Line

`%%bash` must be on the **very first line** of the cell. No Python comments before it!

```python
# This breaks:
# Some comment
%%bash      <-- ERROR: not first line
echo hello

# This works:
%%bash
# Bash comments inside are fine
echo hello
```

### Line Magic Argument Parsing

`%bash` line magic has its own argument parser with flags like `-t`, `-r`, `-s`. This means:

| Command | Works? | Reason |
|---------|--------|--------|
| `%bash pwd` | Yes | No flags in command |
| `%bash echo hello` | Yes | No flags in command |
| `%bash ls -la` | **No** | `-la` is seen as a magic flag |
| `!ls -la` | Yes | `!` passes directly to shell |
| `%%bash` + `ls -la` | Yes | Cell magic doesn't parse flags |

### Bash History Expansion

Avoid `!` in double-quoted strings:
- `echo "Hello!"` - **Error** (bash tries to expand `!"`) 
- `echo 'Hello!'` - **Works** (single quotes prevent expansion)

## Command Reference

| Option | Description |
|--------|-------------|
| `%bash cmd` | Execute simple command (no flags in cmd) |
| `!cmd` | Execute command with shell directly |
| `%%bash` | Multi-line cell magic (must be first line!) |
| `%bash -t SECONDS` | Set timeout (use alone, then run commands) |
| `%bash -r` | Reset the shell interpreter |
| `%bash -s` | Enable sudo mode |
| `%bash -o` | Return ShellInterpreter object |
| `@{var}` | Expand Python variable |

## Tips and Best Practices

1. **Use `!command` for one-liners with flags** - Simplest and most reliable
2. **Use `%%bash` for scripts** - Multi-line commands, supports all flags
3. **Use `%bash` for simple commands** - Quick git status, pwd, echo without flags
4. **`%%bash` on line 1** - Cell magic must be the very first line
5. **Use single quotes** - Avoid bash history expansion issues with `!`
6. **Variable expansion** - Powerful for dynamic commands based on Python analysis
7. **Safe Mode** - Enable for shared notebooks or LLM-generated commands
