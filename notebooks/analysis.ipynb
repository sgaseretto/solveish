## Prerequisites

```python
from my_pkg.helpers import word_count, summarize_counts
```

```python
# Analyze some sample text
text = """
Dialeng is a notebook server that lets you explore ideas interactively.
You can create notebooks, mark cells for export, and reuse your code
across notebooks without any manual copy-paste. The export workflow
extracts marked cells into a _lib directory, making them importable
from any other notebook in the same project. This lets you iterate
quickly: explore in one notebook, export the useful parts, and import
them in the next notebook.
"""

counts = word_count(text)
top_words = summarize_counts(counts, top_n=5)
print("Top 5 words:")
for word, count in top_words:
    print(f"  {word}: {count}")
```

## Adding your own exports

This notebook can also export its own functions. Add `#| default_exp analysis` 
at the top and mark cells with `#| export` to create `my_pkg/analysis.py`.

```python
# You could also chain exports:
#| default_exp analysis  (uncomment to enable export from this notebook)

#| export
from my_pkg.helpers import word_count, summarize_counts

def analyze_file(path: str, top_n: int = 10):
    """Full analysis pipeline: read file -> count words -> summarize."""
    from my_pkg.helpers import read_text
    text = read_text(path)
    counts = word_count(text)
    return summarize_counts(counts, top_n=top_n)
```
