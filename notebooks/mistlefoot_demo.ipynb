# mistlefoot — Extended Markdown Rendering

**mistlefoot** extends standard markdown with subscript, superscript, highlighting, strikethrough, emojis, footnotes, task lists, and heading attributes. Built on top of [mistletoe](https://github.com/miyuchina/mistletoe).

Dialeng uses mistlefoot for both server-side rendering (MIME bundles, API endpoint) and on-demand rendering of note and prompt cells.

## Subscript & Superscript

Use `~text~` for subscript and `^text^` for superscript:

- Water: H~2~O
- Einstein: E=mc^2^
- Chemical formula: Ca~3~(PO~4~)~2~

## Highlighting

Use `==text==` to ==highlight important text==:

- This is ==critically important== information
- You should ==pay attention== to this part

## Strikethrough

Use `~~text~~` for ~~deleted text~~:

- The old API was ~~deprecated~~ removed
- Price: ~~$100~~ $75

## Emojis

Use `:name:` for inline emojis:

- :rocket: Launch ready
- :white_check_mark: Tests passing
- :warning: Careful with this
- :heart: We love markdown
- :star: Five star feature
- :bulb: Good idea
- :fire: Hot take

## Using mistlefoot Programmatically

```python
from mistletoe import markdown
from mistlefoot import ExtendedHtmlRenderer

# Render markdown to HTML
text = "H~2~O is ==water== :droplet:"
html = markdown(text, ExtendedHtmlRenderer)
print(html)
```

```python
# Display rendered HTML in notebook output
from IPython.display import HTML

sample = """
### Features {#features .demo}

- [x] Subscript: H~2~O
- [x] Superscript: E=mc^2^
- [x] ==Highlighting==
- [x] ~~Strikethrough~~
- [x] Emojis :rocket: :star:
"""

HTML(markdown(sample, ExtendedHtmlRenderer))
```

```python
# Compare with basic rendering
print("=== mistlefoot (extended) ===")
print(markdown("H~2~O ==highlighted== :rocket:", ExtendedHtmlRenderer))
print()
print("=== basic mistletoe ===")
from mistletoe import HtmlRenderer
print(markdown("H~2~O ==highlighted== :rocket:", HtmlRenderer))
```

## How It Works in Dialeng

Dialeng uses a **hybrid rendering strategy**:

1. **During editing**: Fast client-side JS rendering for instant preview
2. **After editing/streaming**: Server-side mistlefoot rendering via `/render-markdown` endpoint for full-fidelity output

This means all the extended markdown features above work in note cells and AI response previews automatically!
